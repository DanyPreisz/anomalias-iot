# Detección de Anomalías en Sensores IoT

Proyecto de detección de anomalías listo para **Google Colab**.

**Pregunta:** ¿cómo detectar fallas en sensores IoT?

**Método:** Isolation Forest

**Stack:** pandas, scikit-learn, matplotlib

Datos simulados de sensores.

## 0. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
print("OK")
```

## 1. Generar datos simulados

```python
np.random.seed(42)
n_normal = 500
n_anomalia = 25

normal = np.random.normal([25, 60], [2, 5], (n_normal, 2))
anomalia = np.random.normal([40, 90], [3, 5], (n_anomalia, 2))

X = np.vstack([normal, anomalia])
y = np.array([0]*n_normal + [1]*n_anomalia)

df = pd.DataFrame(X, columns=["temperatura", "humedad"])
df["anomalia"] = y
print(df.shape)
df.head()
```

## 2. Visualización

```python
plt.scatter(df["temperatura"], df["humedad"], c=df["anomalia"], cmap="coolwarm", alpha=0.6)
plt.title("Sensores IoT (rojo = anomalía)")
plt.xlabel("Temperatura")
plt.ylabel("Humedad")
plt.show()
```

## 3. Isolation Forest

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[["temperatura", "humedad"]])

iso = IsolationForest(contamination=0.05, random_state=42)
preds = iso.fit_predict(X_scaled)
df["pred"] = (preds == -1).astype(int)
print(df["pred"].value_counts())
```

## 4. Evaluación

```python
from sklearn.metrics import classification_report
print(classification_report(df["anomalia"], df["pred"]))
```

## 5. Conclusiones

- Isolation Forest es útil para monitoreo de sensores IoT.
- Detecta fallas o comportamientos anómalos sin etiquetas.
- Se puede extender a series temporales o más sensores.
